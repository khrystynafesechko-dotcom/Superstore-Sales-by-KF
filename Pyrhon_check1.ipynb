```python
Store_Sales["sales"] = pd.to_numeric(Store_Sales["sales"].astype(str).str.replace(",", "", regex=False),errors="coerce")
print(Store_Sales["sales"].sum())
print(Store_Sales["sales"].isna().sum())
```

```python
total_sales = Store_Sales["sales"].sum()
total_profit = Store_Sales["profit"].sum()
margin = total_profit / total_sales
orders = Store_Sales["order_id"].nunique()
customers = Store_Sales["customer_name"].nunique()

print("Total Sales:", round(total_sales))
print("Total Profi:", round(total_profit, 2))
print("Profit Margin:", round(margin * 100, 2), "%")
print("Total Orders:", orders)
print("Total Customers:", customers)
```

```python
by_year = Store_Sales.groupby("year")[["sales", "profit"]].sum()
by_year["margin_%"] = by_year["profit"] / by_year["sales"] * 100
by_year["YoY_%"] = by_year["sales"].pct_change() * 100
by_year.round(1)
```
